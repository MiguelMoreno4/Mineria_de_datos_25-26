Como ves, puede haber problemas a la hora de encontrarnos caracteres inesperados, mejor una ER

```python
res = collection.update_many(
 {"RECORD": {"$type": "string"}},
 [{"$set": {
 "match": {"$regexFind": {
 "input": "$RECORD",
 "regex": "^(\\d+).*?-(\\d+).*?-(\\d+)"
 }}
 }},
 {"$set": {
 "WINS": {
 "$cond": {
 "if": {"$ne": ["$match", None]},
 "then": {"$toInt": {"$arrayElemAt": ["$match.captures", 0]}},
 "else": 0
 }
 }
 }},
 {"$unset": ["match"]} # Borra el temporal
 ]
)

print(res) 
```

```python
#Otro elemento más es `$function`, el cual permite ejecutar codigo javascript dentro de la consulta. Esto es MUY lento, no se suele utilizar salvo
#que sea indispensable.
from pymongo import MongoClient
from bson.code import Code 

pipeline = [
 {
 "$set": {
 "WEIGHT": {
 "$function": {
 "body": Code("""
 function (w) {
 if (w == null) return null;
 const m = String(w).match(/\\d+/);
 return m ? parseInt(m[0], 10) : null;
 }
 """),
 "args": ["$WEIGHT"],
 "lang": "js"
 }
 }
 }
 }
]

res = collection.update_many({"WEIGHT": {"$type": "string"}}, pipeline)
print(res.modified_count, "documentos actualizados")
```
